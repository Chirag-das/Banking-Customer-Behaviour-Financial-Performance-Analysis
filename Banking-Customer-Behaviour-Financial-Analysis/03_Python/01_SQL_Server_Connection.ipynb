```python

# ============================================================
# BANKING CUSTOMER BEHAVIOUR & FINANCIAL PERFORMANCE ANALYSIS
# SQL SERVER CONNECTION & DATA LOADING
# ============================================================

import pandas as pd
import numpy as np
import sqlalchemy
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# ------------------------------------------------------------
# 1. LIBRARY VERSION CHECK
# ------------------------------------------------------------

print("=" * 55)
print("LIBRARY VERSIONS")
print("=" * 55)

print("Pandas Version:", pd.__version__)
print("NumPy Version:", np.__version__)
print("SQLAlchemy Version:", sqlalchemy.__version__)


```

```python
# ------------------------------------------------------------
# 2. DATABASE CONFIGURATION
# ------------------------------------------------------------

server = r"DESKTOP-TL9ULDS\SQLEXPRESS"
database = "Banking_Analytics"
driver = "ODBC Driver 18 for SQL Server"

connection_url = URL.create(
    "mssql+pyodbc",
    host=server,
    database=database,
    query={
        "driver": driver,
        "trusted_connection": "yes",
        "TrustServerCertificate": "yes"
    }
)

engine = create_engine(connection_url)

print("\nDatabase configuration completed!")
```

```python
# ------------------------------------------------------------
# 3. TEST DATABASE CONNECTION
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("DATABASE CONNECTION")
print("=" * 55)

with engine.connect() as connection:
    database_name = connection.exec_driver_sql(
        "SELECT DB_NAME()"
    ).scalar_one()

    print("SQL Server Connection Successful!")
    print("Connected Database:", database_name)
```

```python
# ------------------------------------------------------------
# 4. LOAD CLEANED DATASETS
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("LOADING CLEANED DATASETS")
print("=" * 55)

with engine.connect() as connection:

    bank_df = pd.read_sql_query(
        "SELECT * FROM dbo.Bank_Clean",
        con=connection
    )

    customer_df = pd.read_sql_query(
        "SELECT * FROM dbo.Customer_Clean",
        con=connection
    )

    transaction_df = pd.read_sql_query(
        "SELECT * FROM dbo.Transactions_Clean",
        con=connection
    )

print("All cleaned datasets loaded successfully!")
```

```python
# ------------------------------------------------------------
# 5. DATASET DIMENSIONS
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("DATASET DIMENSIONS")
print("=" * 55)

print("Bank Dataset:", bank_df.shape)
print("Customer Dataset:", customer_df.shape)
print("Transaction Dataset:", transaction_df.shape)
```

```python
# ------------------------------------------------------------
# 6. DATASET PREVIEW
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("BANK DATA PREVIEW")
print("=" * 55)
display(bank_df.head())

print("\n" + "=" * 55)
print("CUSTOMER DATA PREVIEW")
print("=" * 55)
display(customer_df.head())

print("\n" + "=" * 55)
print("TRANSACTION DATA PREVIEW")
print("=" * 55)
display(transaction_df.head())
```

```python
# ------------------------------------------------------------
# 7. DATA TYPES
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("DATA TYPES")
print("=" * 55)

print("\nBank Data Types:")
print(bank_df.dtypes)

print("\nCustomer Data Types:")
print(customer_df.dtypes)

print("\nTransaction Data Types:")
print(transaction_df.dtypes)
```

```python
# ------------------------------------------------------------
# 8. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("DATABASE CONNECTION & DATA LOADING COMPLETED")
print("=" * 55)

# Close SQLAlchemy engine after loading
engine.dispose()
```
